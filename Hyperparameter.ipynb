```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

x = np.random.rand(100, 4)
y = np.random.randint(0, 2, 100)

for lr in [0.01, 0.001]:
    for units in [8, 16]:
        m = Sequential([
            Dense(units, activation='relu', input_shape=(4,)),
            Dense(1, activation='sigmoid')
        ])

        m.compile(optimizer=Adam(learning_rate=lr),
                  loss='binary_crossentropy', metrics=['accuracy'])

        m.fit(x, y, epochs=5, verbose=0)
        loss, acc = m.evaluate(x, y, verbose=0)

        print("LR:", lr, "Units:", units,
              "Loss:", round(loss, 4), "Accuracy:", round(acc, 4))
```
