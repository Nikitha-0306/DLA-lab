```python
from gensim.models import Word2Vec

sentences = [
    ["i", "love", "machine", "learning"],
    ["i", "love", "deep", "learning"],
    ["machine", "learning", "is", "useful"]
]

# Train the Word2Vec model
model = Word2Vec(sentences, vector_size=10, window=2, min_count=1, sg=1, epochs=50)

# Get the vector representation of "learning"
print("Vector for 'learning':")
print(model.wv["learning"])

# Find words most similar to "learning"
print("\nWords most similar to 'learning':")
print(model.wv.most_similar("learning"))
```
