```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

texts = ["I love this movie", "This movie is good",
         "I hate this movie", "This is a bad movie"]
y = np.array([1, 1, 0, 0])

tok = Tokenizer(num_words=1000)
tok.fit_on_texts(texts)
x = tok.texts_to_sequences(texts)
x = pad_sequences(x, maxlen=8)

m = Sequential([
    Embedding(1000, 32, input_length=8),
    LSTM(32),
    Dense(1, activation='sigmoid')
])

m.compile(optimizer='adam', loss='binary_crossentropy',
          metrics=['accuracy'])
m.fit(x, y, epochs=20, verbose=0)

loss, acc = m.evaluate(x, y, verbose=0)
print("Loss:", loss)
print("Accuracy:", acc)

p = m.predict(x, verbose=0)
for i in range(len(x)):
    print("Actual:", y[i], "Predicted:", round(p[i][0]))
```
